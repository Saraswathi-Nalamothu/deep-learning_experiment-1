```python
import numpy as np
import matplotlib.pyplot as plt
import os

os.makedirs("eps_plots", exist_ok=True)
```

```python
def perceptron_train(X, y, name, lr=0.1, epochs=10):

    w = np.zeros(X.shape[1])
    b = 0
    update = 1

    for epoch in range(epochs):
        error = 0

        for i in range(len(X)):

            result = np.dot(X[i], w) + b

            pred = 1 if result >= 0 else 0

            if pred != y[i]:

                w = w + lr * (y[i]-pred) * X[i]
                b = b + lr * (y[i]-pred)

                error += 1

                print(name, "Update", update)
                print("Weights:", w)
                print("Bias:", b)

                if X.shape[1] == 2:
                    plot_boundary(X,y,w,b,name,update)

                update += 1

        if error == 0:
            break

    return w,b
```

```python
def plot_boundary(X,y,w,b,name,count):

    plt.figure(figsize=(5,4))

    for i in range(len(X)):
        if y[i]==0:
            plt.scatter(X[i,0],X[i,1],marker="o")
        else:
            plt.scatter(X[i,0],X[i,1],marker="x")

    x1=np.linspace(-1,2,100)

    if w[1]!=0:
        x2=-(w[0]*x1+b)/w[1]
        plt.plot(x1,x2)

    plt.xlabel("X1")
    plt.ylabel("X2")
    plt.title(name+" Update "+str(count))

    plt.savefig(
        "eps_plots/"+name+"_update_"+str(count)+".eps",
        format="eps"
    )

    plt.show()
```

```python
X_and=np.array([
[0,0],
[0,1],
[1,0],
[1,1]
])

y_and=np.array([0,0,0,1])

w_and,b_and=perceptron_train(
    X_and,
    y_and,
    "AND"
)

print("Final weights:",w_and)
print("Final bias:",b_and)
```

```python
X_or=np.array([
[0,0],
[0,1],
[1,0],
[1,1]
])

y_or=np.array([0,1,1,1])

w_or,b_or=perceptron_train(
    X_or,
    y_or,
    "OR"
)

print("Final weights:",w_or)
print("Final bias:",b_or)
```

```python
X_not=np.array([
[0],
[1]
])

y_not=np.array([
1,
0
])

w_not,b_not=perceptron_train(
    X_not,
    y_not,
    "NOT"
)

print("Final weights:",w_not)
print("Final bias:",b_not)
```

```python
X_xor=np.array([
[0,0],
[0,1],
[1,0],
[1,1]
])

y_xor=np.array([
0,
1,
1,
0
])

w_xor,b_xor=perceptron_train(
    X_xor,
    y_xor,
    "XOR",
    epochs=20
)

print("Final weights:",w_xor)
print("Final bias:",b_xor)
```
