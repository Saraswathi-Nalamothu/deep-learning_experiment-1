```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([[0],[1],[1],[0]])

input_neurons = 2
hidden_neurons = 4
output_neurons = 1

lr = 0.5
epochs = 100

W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.random.randn(1, hidden_neurons)

W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.random.randn(1, output_neurons)

print("\nInitial Weights and Biases\n")

print("W1")
print(W1)

print("\nb1")
print(b1)

print("\nW2")
print(W2)

print("\nb2")
print(b2)

def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

def plot_boundary(W1,b1,W2,b2,title):

    h = 0.02

    x_min, x_max = -0.5, 1.5
    y_min, y_max = -0.5, 1.5

    xx, yy = np.meshgrid(np.arange(x_min,x_max,h),
                         np.arange(y_min,y_max,h))

    grid = np.c_[xx.ravel(),yy.ravel()]

    hidden = sigmoid(np.dot(grid,W1)+b1)
    output = sigmoid(np.dot(hidden,W2)+b2)

    Z = (output>=0.5).astype(int)
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(5,5))
    plt.contourf(xx,yy,Z,alpha=0.4)

    plt.scatter(X[:,0],X[:,1],c=y.ravel(),s=150,edgecolors='k')

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)
    plt.title(title)
    plt.xlabel("Input 1")
    plt.ylabel("Input 2")
    plt.grid(True)
    plt.show()

for epoch in range(epochs):

    hidden_input = np.dot(X,W1)+b1
    hidden_output = sigmoid(hidden_input)

    output_input = np.dot(hidden_output,W2)+b2
    predicted = sigmoid(output_input)

    loss = np.mean((y-predicted)**2)

    output_error = y-predicted
    output_delta = output_error*sigmoid_derivative(predicted)

    hidden_error = np.dot(output_delta,W2.T)
    hidden_delta = hidden_error*sigmoid_derivative(hidden_output)

    old_W1 = W1.copy()
    old_W2 = W2.copy()
    old_b1 = b1.copy()
    old_b2 = b2.copy()

    W2 += lr*np.dot(hidden_output.T,output_delta)
    b2 += lr*np.sum(output_delta,axis=0,keepdims=True)

    W1 += lr*np.dot(X.T,hidden_delta)
    b1 += lr*np.sum(hidden_delta,axis=0,keepdims=True)

    print("="*60)
    print("Epoch",epoch+1)
    print("Loss =",loss)

    print("\nChange in W1")
    print(W1-old_W1)

    print("\nChange in b1")
    print(b1-old_b1)

    print("\nChange in W2")
    print(W2-old_W2)

    print("\nChange in b2")
    print(b2-old_b2)

    plot_boundary(W1,b1,W2,b2,f"Decision Boundary - Epoch {epoch+1}")

print("\nTraining Completed")

print("\nFinal W1")
print(W1)

print("\nFinal b1")
print(b1)

print("\nFinal W2")
print(W2)

print("\nFinal b2")
print(b2)

hidden = sigmoid(np.dot(X,W1)+b1)
output = sigmoid(np.dot(hidden,W2)+b2)

predictions = (output>=0.5).astype(int)

accuracy = np.mean(predictions==y)*100

print("\nPrediction Table\n")

print("Input1\tInput2\tActual\tPredicted")

for i in range(len(X)):
    print(
        X[i][0],
        "\t",
        X[i][1],
        "\t",
        y[i][0],
        "\t",
        predictions[i][0]
    )

print("\nAccuracy =",accuracy,"%")

plot_boundary(W1,b1,W2,b2,"Final Decision Boundary")
```
